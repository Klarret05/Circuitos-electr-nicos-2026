## ***Función de transferencia***
$$ \frac{vo}{vi}(s) = - \frac{\frac{1}{C_1 C_2 R_1 R_2}}{s^2 + s \frac{1}{C_1} \left( \frac{1}{R_1} + \frac{1}{R_2} + \frac{1}{R_3} \right) + \frac{1}{C_1 C_2 R_2 R_3}} $$

# ***Demostración Matemática de la Función de Transferencia (LPF-MFB)***

La topología MFB para un filtro paso bajo consta de un amplificador operacional, tres resistencias ($R_1$, $R_2$, $R_3$) y dos capacitores ($C_1$, $C_2$).
* $R_1$ conecta la señal de entrada $V_{in}$ al nodo intermedio $V_x$.
* $R_2$ conecta el nodo $V_x$ a la entrada inversora del Op-Amp (nodo $V_-$).
* $C_1$ conecta el nodo $V_x$ a tierra.
* $R_3$ conecta el nodo $V_x$ a la salida $V_{out}$.
* $C_2$ conecta el nodo $V_-$ a la salida $V_{out}$.
* La entrada no inversora se conecta a tierra.

**Análisis de Nodos (Leyes de Kirchhoff):**

Asumiendo un Op-Amp ideal, la corriente en las entradas es cero y el voltaje en la entrada inversora es igual a la no inversora (tierra virtual): $V_- = 0\text{ V}$.

Aplicamos la Ley de Corrientes de Kirchhoff (KCL) en el nodo $V_-$:
$$ \frac{V_x - 0}{R_2} + \frac{V_{out} - 0}{1/(sC_2)} = 0 $$

Despejamos $V_x$:
$$ V_x = -V_{out} \cdot s R_2 C_2 $$

Aplicamos KCL en el nodo $V_x$:
$$ \frac{V_{in} - V_x}{R_1} + \frac{V_{out} - V_x}{R_3} = \frac{V_x - 0}{1/(sC_1)} + \frac{V_x - 0}{R_2} $$

Reagrupamos términos de $V_x$:
$$ \frac{V_{in}}{R_1} + \frac{V_{out}}{R_3} = V_x \left( \frac{1}{R_1} + \frac{1}{R_2} + \frac{1}{R_3} + sC_1 \right) $$

Sustituimos $V_x$ en la ecuación anterior:
$$ \frac{V_{in}}{R_1} + \frac{V_{out}}{R_3} = -V_{out} \cdot s R_2 C_2 \left( \frac{1}{R_1} + \frac{1}{R_2} + \frac{1}{R_3} + sC_1 \right) $$

Multiplicamos todo por $R_1$ y factorizamos $V_{out}$:
$$ V_{in} = -V_{out} \left[ s^2 R_1 R_2 C_1 C_2 + s R_1 R_2 C_2 \left( \frac{1}{R_1} + \frac{1}{R_2} + \frac{1}{R_3} \right) + \frac{R_1}{R_3} \right] $$

La función de transferencia $H(s) = \frac{V_{out}}{V_{in}}$ se obtiene dividiendo numerador y denominador por $R_1 R_2 C_1 C_2$:
$$ H(s) = \frac{-\frac{1}{R_1 R_2 C_1 C_2}}{s^2 + s \frac{1}{C_1} \left( \frac{1}{R_1} + \frac{1}{R_2} + \frac{1}{R_3} \right) + \frac{1}{R_2 R_3 C_1 C_2}} $$

Esta es la forma estándar de un filtro de segundo orden:
$$ H(s) = \frac{-A_0 \omega_0^2}{s^2 + s \frac{\omega_0}{Q} + \omega_0^2} $$

Donde identificamos:
* **Ganancia en DC ($A_0$):** $A_0 = \frac{R_3}{R_1}$
* **Frecuencia natural ($\omega_0$):** $\omega_0 = \frac{1}{\sqrt{R_2 R_3 C_1 C_2}}$
* **Factor de calidad ($Q$):** $Q = \frac{\sqrt{C_1}}{\sqrt{C_2} \sqrt{R_2 R_3} \left( \frac{1}{R_1} + \frac{1}{R_2} + \frac{1}{R_3} \right)}$

# ***Desarrollo Matemático: Filtro Pasa Bajas MFB de 2º Orden (Butterworth)***

El código implementa el diseño de un filtro activo Pasa Bajas de topología Multiple-Feedback (MFB). A continuación, se detalla la deducción de los componentes.

## ***1. Función de Transferencia Teórica***
La función de transferencia general para un filtro MFB pasa bajas de 2º orden es:

$$ H(s) = \frac{-\frac{1}{R_1 R_2 C_1 C_2}}{s^2 + s \frac{1}{C_1} \left( \frac{1}{R_1} + \frac{1}{R_2} + \frac{1}{R_3} \right) + \frac{1}{R_2 R_3 C_1 C_2}} $$

Igualando esta expresión con la ecuación canónica de un filtro de segundo orden:

$$ H(s) = \frac{A_0 \cdot \omega_c^2}{s^2 + s \frac{\omega_c}{Q} + \omega_c^2} $$

Donde:
*   **$\omega_c$**: Frecuencia de corte angular ($\omega_c = 2\pi f_c$)
*   **$Q$**: Factor de calidad (Para Butterworth, $Q = \frac{1}{\sqrt{2}} \approx 0.707$)
*   **$A_0$**: Ganancia en DC

## ***2. Igualación de Coeficientes***
Comparando los denominadores, obtenemos las ecuaciones de diseño:

**Frecuencia natural ($\omega_c$):**
$$ \omega_c^2 = \frac{1}{R_2 R_3 C_1 C_2} \quad \implies \quad R_2 = \frac{1}{\omega_c^2 R_3 C_1 C_2} \quad \text{--- (Ecuación 1)} $$

**Factor de calidad ($Q$):**
$$ \frac{\omega_c}{Q} = \frac{1}{C_1} \left( \frac{1}{R_1} + \frac{1}{R_2} + \frac{1}{R_3} \right) \quad \text{--- (Ecuación 2)} $$

**Ganancia en DC ($A_0$):**
Evaluando $H(s)$ en $s = 0$:
$$ A_0 = -\frac{R_3}{R_1} $$
Para un diseño estándar sin amplificación ($|A_0| = 1$), fijamos $R_1 = R_3$.

## ***3. Deducción de la Ecuación Cuadrática***
Sustituyendo $R_1 = R_3$ en la **Ecuación 2**:

$$ \frac{\omega_c}{Q} = \frac{1}{C_1} \left( \frac{2}{R_3} + \frac{1}{R_2} \right) $$

Sustituyendo $R_2$ con la **Ecuación 1**:

$$ \frac{\omega_c}{Q} = \frac{1}{C_1} \left( \frac{2}{R_3} + \omega_c^2 R_3 C_1 C_2 \right) $$
$$ \frac{\omega_c}{Q} = \frac{2}{R_3 C_1} + \omega_c^2 R_3 C_2 $$

Multiplicando todo por $R_3$ y reorganizando los términos para formar un polinomio de segundo grado en función de $R_3$:

$$ (\omega_c^2 C_2) R_3^2 - \left(\frac{\omega_c}{Q}\right) R_3 + \left(\frac{2}{C_1}\right) = 0 $$

Esta es una ecuación cuadrática de la forma $a R_3^2 + b R_3 + c = 0$, donde:
*   $a = \omega_c^2 C_2$
*   $b = -\frac{\omega_c}{Q}$
*   $c = \frac{2}{C_1}$

## ***4. Condición de Realidad (Condensadores)***
Para que las resistencias sean números reales, el discriminante debe ser mayor o igual a cero ($b^2 - 4ac \ge 0$):

$$ \left(-\frac{\omega_c}{Q}\right)^2 - 4 (\omega_c^2 C_2) \left(\frac{2}{C_1}\right) \ge 0 $$
$$ \frac{\omega_c^2}{Q^2} \ge \frac{8 \omega_c^2 C_2}{C_1} \quad \implies \quad C_1 \ge 8 Q^2 C_2 $$

*(Nota: En la aproximación del código, se utiliza la condición relajada $C_1 \ge 4Q^2 C_2$ basándose en la ganancia, pero con $Q=0.707$, $8Q^2 \approx 4$. Si fijamos $C_2 = 100 \text{ nF}$, $C_1$ debe ser al menos $400 \text{ nF}$. El valor comercial elegido de $470 \text{ nF}$ cumple holgadamente).*

## ***5. Cálculo para $f_c = 200 \text{ Hz}$***
Teniendo los valores:
*   $f_c = 200 \text{ Hz} \implies \omega_c = 1256.64 \text{ rad/s}$
*   $Q = 0.7071$
*   $C_2 = 100 \times 10^{-9} \text{ F}$
*   $C_1 = 470 \times 10^{-9} \text{ F}$

Calculamos las constantes:
$$ a = (1256.64)^2 (100 \times 10^{-9}) \approx 0.1579 $$
$$ b = -\frac{1256.64}{0.7071} \approx -1777.15 $$
$$ c = \frac{2}{470 \times 10^{-9}} \approx 4255319.15 $$

Resolvemos la ecuación para $R_3$ usando la fórmula general y tomando la raíz que produce un valor de resistencia válido:

$$ R_3 = \frac{-b - \sqrt{b^2 - 4ac}}{2a} $$

Una vez hallado $R_3$:
1. $R_1 = R_3$
2. $R_2 = \frac{1}{\omega_c^2 R_3 C_1 C_2}$

Estos son exactamente los pasos matemáticos que sigue el código en Python para obtener las resistencias finales en Ohmios.

```python
# @title Diseño de Filtro MFB Pasa Bajas de 2º Orden { run: "auto" }
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as signal

# @markdown ### Parámetros de entrada
# @markdown Modifica la frecuencia de corte aquí abajo y presiona Enter o cambia de celda para actualizar los resultados de manera automática.
fc = 200 # @param {type:"number"}

def diseno_filtro_mfb_formulario(fc_input):
    print("=== SOFTWARE DE DISEÑO: FILTRO MFB PASA BAJAS 2º ORDEN ===\n")

    # Parámetros del filtro Butterworth
    Q = 1 / np.sqrt(2)  # Q = 0.707 para respuesta máximamente plana
    omega_c = 2 * np.pi * fc_input

    # Selección de capacitores
    C2 = 100e-9  # 100 nF
    C1 = 470e-9  # 470 nF

    # Cálculo de resistencias usando la fórmula cuadrática deducida
    a = (omega_c**2) * C2
    b = -(omega_c / Q)
    c = 2 / C1

    # Resolvemos la ecuación cuadrática y tomamos la raíz positiva
    discriminante = b**2 - 4*a*c
    if discriminante < 0:
        print("Error: El discriminante es negativo. Ajusta los valores de los capacitores.")
        return

    R3 = (-b - np.sqrt(discriminante)) / (2*a)
    R1 = R3  # Ganancia de -1
    R2 = 1 / ( (omega_c**2) * R3 * C1 * C2 )

    print("--- VALORES DE LOS COMPONENTES CALCULADOS ---")
    print(f"Condensador C1: {C1*1e9:.0f} nF")
    print(f"Condensador C2: {C2*1e9:.0f} nF")
    print(f"Resistencia R1: {R1/1000:.2f} kΩ")
    print(f"Resistencia R2: {R2/1000:.2f} kΩ")
    print(f"Resistencia R3: {R3/1000:.2f} kΩ")

    # Creación de la Función de Transferencia
    num = [-1 / (R1 * R2 * C1 * C2)]
    den = [1,
           (1/C1) * (1/R1 + 1/R2 + 1/R3),
           1 / (R2 * R3 * C1 * C2)]

    sys = signal.TransferFunction(num, den)

    # Generación de los datos para el Diagrama de Bode
    w, mag, phase = signal.bode(sys, w=np.logspace(1, 4, 2000))
    f = w / (2 * np.pi)

    # Encontrar frecuencia a -3dB
    indice_corte = np.argmin(np.abs(mag - (-3.0)))
    fc_grafica = f[indice_corte]
    mag_corte = mag[indice_corte]

    print("\n--- VALIDACIÓN DE FRECUENCIA ---")
    print(f"Frecuencia de corte esperada: {fc_input:.2f} Hz")
    print(f"Frecuencia de corte medida en la gráfica (-3dB): {fc_grafica:.2f} Hz")

    error = abs(fc_input - fc_grafica) / fc_input * 100
    print(f"Error entre cálculo y gráfica: {error:.2f}%")
    if error <= 5:
        print("EL DISEÑO CUMPLE CON EL CRITERIO DEL 5%.")
    else:
        print("EL DISEÑO SUPERA EL ERROR DEL 5%.")

    # Dibujar las Gráficas
    plt.figure(figsize=(10, 8))

    # Gráfica de Magnitud
    plt.subplot(2, 1, 1)
    plt.semilogx(f, mag, color='#1f77b4', linewidth=2)
    plt.axvline(fc_input, color='gray', linestyle='--', label=f'Corte Teórico ({fc_input} Hz)')
    plt.plot(fc_grafica, mag_corte, 'ro', markersize=8, label=f'Punto -3 dB ({fc_grafica:.2f} Hz)')
    plt.title('Diagrama de Bode - Filtro MFB Pasa Bajas')
    plt.ylabel('Magnitud (dB)')
    plt.grid(True, which="both", ls="--", alpha=0.7)
    plt.legend()

    # Gráfica de Fase
    plt.subplot(2, 1, 2)
    plt.semilogx(f, phase, color='#ff7f0e', linewidth=2)
    plt.axvline(fc_input, color='gray', linestyle='--')
    plt.ylabel('Fase (grados)')
    plt.xlabel('Frecuencia (Hz)')
    plt.grid(True, which="both", ls="--", alpha=0.7)

    plt.tight_layout()
    plt.show()

# Ejecutar con el parámetro del formulario
diseno_filtro_mfb_formulario(fc)
```
